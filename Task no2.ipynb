# Task 2: Data Cleaning and Preprocessing

## Objective
The objective of this task is to clean and preprocess the Titanic dataset by handling missing values, removing duplicate records, detecting and treating outliers, and normalizing numerical features. This improves data quality and prepares the dataset for analysis and machine learning applications.

Load the Titanic dataset into a Pandas DataFrame for further analysis and preprocessing.

## Check Missing Values

Identify missing values present in each column of the dataset.

```python
df.isnull().sum()
```

## Handle Missing Values

Missing values are treated using appropriate techniques:
- Age is filled using the median value.
- Embarked is filled using the mode value.
- Cabin is removed because it contains a large number of missing values.

```python
# Fill missing values in Age using median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing values in Embarked using mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin column due to excessive missing values
df.drop("Cabin", axis=1, inplace=True)

print("Missing values handled successfully")
```

## Verify Missing Values

Confirm that all missing values have been handled successfully.

```python
df.isnull().sum()
```

## Check Duplicate Records

Identify duplicate rows present in the dataset.

```python
df.duplicated().sum()
```

## Detect Outliers

The Interquartile Range (IQR) method is used to identify outliers in the Fare column.

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

## Handle Outliers

Records containing Fare values outside the acceptable IQR range are removed.

```python
original_shape = df.shape

df = df[(df["Fare"] >= lower_bound) & (df["Fare"] <= upper_bound)]

print("Original Shape:", original_shape)
print("New Shape:", df.shape)
```

## Normalize Features

Numerical features are scaled using MinMaxScaler to bring values into a common range between 0 and 1.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Age", "Fare"]] = scaler.fit_transform(df[["Age", "Fare"]])

df.head()
```

## Save Clean Dataset

The cleaned and preprocessed dataset is saved as a CSV file for future analysis.

```python
df.to_csv("cleaned_titanic.csv", index=False)

print("Cleaned dataset saved successfully!")
```
